# Importing Modules

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings 
warnings.filterwarnings('ignore')
```

# Loding Data

```python
df=pd.read_csv("/kaggle/input/datasets/karkavelrajaj/amazon-sales-dataset/amazon.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
print(f"Rows:{df.shape[0]},Columns:{df.shape[1]}")
```

```python
print(f"No of duplicate rows: {df.duplicated().sum()}")
```

```python
df["discounted_price"] = (
    df["discounted_price"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)

df["actual_price"] = (
    df["actual_price"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)

df['discount_percentage']=df['discount_percentage'].str.replace("%","").str.strip().astype(int)

df['rating_count']=df['rating_count'].str.replace(",","")
df['rating_count']=df['rating_count'].apply(pd.to_numeric,errors='coerce')

df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
df['main_category'] = df['category'].str.split('|').str[0]
```

```python
df.isna().sum()
```

```python
for col in df.columns:
    print(f"{col}:{df[col].nunique()}")
```

```python
df['rating_count'].fillna(df['rating_count'].mean(),inplace=True)
```

```python
df['rating'].fillna(df['rating'].mean(),inplace=True)
```

```python
for col in df.columns:
    print(f"{col}:{df[col].nunique()}")
```

# EDA

```python
numeric_columns=['discounted_price','actual_price','discount_percentage',
                "rating",'rating_count']
sns.set_theme()

for col in numeric_columns:
    sns.histplot(x=df[col],kde=True,bins=30)
    plt.show()
```

* mostly sold products price is way lesser than 20000 rupees.
* mostly products are rated from 4 to 4.5.

```python
sns.barplot(y=df['main_category'].value_counts().head(5).index,
           x=df['main_category'].value_counts().head(5).values,)
plt.title("Most product sold by Categories")
plt.ylabel("Category")
plt.xlabel("Count")
plt.show()
```

**mostly product sold are of category of Electronics,computers Accessories and home&kitchen.**

```python
sns.regplot(x=df['discounted_price'],y=df['actual_price'])
plt.title("actual_prics vs discounted_price")
plt.show()
```

**There is a linear releation between actual and discounted price which is obvious**

```python
plt.figure(figsize=(10,5))
sns.scatterplot(x=df['discount_percentage'],y=df['actual_price'],hue=df['main_category'],alpha=0.5)
plt.title("actual_prics vs discount_percentage")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(x='actual_price',y='rating',data=df,color="Green",alpha=0.5)
plt.title("rating vs actual_price")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(x='discount_percentage',y='rating',data=df,color="Orange",alpha=0.5)
plt.title("rating vs discount_percentage")
plt.show()
```
